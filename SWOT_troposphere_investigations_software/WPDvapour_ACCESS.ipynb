```python
import numpy as np
import xarray as xr
from ZWD_functions import *
from metpy.calc import mixing_ratio_from_relative_humidity
from metpy.calc import specific_humidity_from_mixing_ratio
from metpy.units import units
```

```python
### Bass Strait region

#define start and end date
start_date = "20230330"
end_date = "20230710"

# define directory
directory_path = './'

all_dates = date_range_subdirs(start_date, end_date, directory_path)

print(all_dates)

ZWD_combined = []
time_combined = []

for i in all_dates: 
    yyyymmdd = str(i)

    try:
        # open datasets (air temp, relative humidity, and geopotential height
        ds_air_temp = xr.open_mfdataset(directory_path + str(yyyymmdd) + '/*/an/pl/air_temp.nc')
        ds_relhum = xr.open_mfdataset(directory_path + str(yyyymmdd) + '/*/an/pl/relhum.nc')
        ds_geop_ht = xr.open_mfdataset(directory_path + str(yyyymmdd) + '/*/an/pl/geop_ht.nc')
        print('opened ', yyyymmdd) 
        
    except ValueError as e:
        print('Error opening ', yyyymmdd) 
        
    # trim to region
    min_lon = 142
    max_lon = 147
    min_lat = -42
    max_lat = -38

    ds_air_temp = ds_air_temp.sel(lat=slice(max_lat, min_lat), lon=slice(min_lon, max_lon))
    ds_relhum = ds_relhum.sel(lat=slice(max_lat, min_lat), lon=slice(min_lon, max_lon))    
    
    p = ds_relhum.lvl.values
    p = p.reshape(27, 1, 1, 1)
    p = p * np.ones((1, len(ds_relhum.lat), len(ds_relhum.lon), len(ds_relhum.time)))
    p = np.transpose(p, (3, 0, 1, 2))
    p = p * units.hPa
    T = ds_air_temp.air_temp.values * units.degK
    rh = ds_relhum.relhum.values * units.percent
    mixing_ratio = mixing_ratio_from_relative_humidity(p, T, rh).to('g/kg')
    spec_hum = specific_humidity_from_mixing_ratio(mixing_ratio* units('g/kg'))
    ZWD = (1.116454e-3 * np.trapz(spec_hum.magnitude/1000, p.magnitude, axis=1) + 17.66543928 * np.trapz((spec_hum.magnitude/1000)/ds_air_temp.air_temp.values, p.magnitude, axis=1)) * (1 + 0.0026*np.cos(2*ds_relhum.lat.values[0]))
    ZWD = xr.DataArray(ZWD, dims=('time', 'lat', 'lon'), coords={'time': ds_air_temp.time, 'lat': ds_air_temp.lat, 'lon': ds_air_temp.lon}, name='ZWD')
    ZWD_combined.append(ZWD)

ZWD_all =  xr.concat(ZWD_combined, dim='time')
ZWD_all.to_netcdf('ACCESS_ZWD_bst_20230330_20230710.nc')
```
