```python
import numpy as np
import xarray as xr
from ZWD_functions import *
```

```python
### Bass Strait region

#define start and end date
start_date = "20230330"
end_date = "20230710"

# define directory
directory_path = './'

all_dates = date_range_subdirs(start_date, end_date, directory_path)

print(all_dates)

ZWD_combined = []
time_combined = []

for i in all_dates: 
    yyyymmdd = str(i)

    try:
        # open datasets 
        ds_cloud_water = xr.open_mfdataset(directory_path + yyyymmdd + '/*/an/ml/cld_water.nc')
        ds_air_temp = xr.open_mfdataset(directory_path + yyyymmdd + '/*/an/ml/air_temp.nc')
        ds_pressure = xr.open_mfdataset(directory_path + yyyymmdd + '/*/an/ml/pressure.nc')
        print('opened ', yyyymmdd) 
        
    except ValueError as e:
        print('Error opening ', yyyymmdd) 
        
    # trim to region
    min_lon = 142
    max_lon = 147
    min_lat = -42
    max_lat = -38

    ds_cloud_water = ds_cloud_water.sel(lat=slice(max_lat, min_lat), lon=slice(min_lon, max_lon))
    ds_air_temp = ds_air_temp.sel(lat=slice(max_lat, min_lat), lon=slice(min_lon, max_lon))
    ds_pressure = ds_pressure.sel(lat=slice(max_lat, min_lat), lon=slice(min_lon, max_lon))   
    
    # specific gass constant for dry air
    R = 287.05

    # air density
    air_density = ds_pressure.pressure/(ds_air_temp.air_temp*R)

    liquid_water_density = ds_cloud_water.cld_water * air_density
    liquid_water_path = np.trapz(liquid_water_density, ds_cloud_water.theta_lvl, axis = 1) # divide by 1000 (density of water) to get from kg/m^2 to m. Multiply by 1000 to get from m to mm. 
    
    ZWD = liquid_water_path * 1.6
    
    print("Max ZWDliq this day = %.1f mm" % np.nanmax(ZWD))
    
    ZWD = xr.DataArray(ZWD, dims=('time', 'lat', 'lon'), coords={'time': liquid_water_density.time, 'lat': ds_air_temp.lat, 'lon': ds_air_temp.lon}, name='ZWD_liq')
    
    ZWD_combined.append(ZWD)
    
ZWD_all =  xr.concat(ZWD_combined, dim='time')
ZWD_all.to_netcdf('ACCESS_ZWD_liq_bst_20230330_20230710.nc')
```
